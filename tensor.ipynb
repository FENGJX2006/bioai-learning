```python
import torch

# 一条"蛋白序列"的简化表示：5个氨基酸 × 4维特征
x = torch.tensor([[0.1, 0.5, 0.2, 0.8],
                  [0.3, 0.7, 0.1, 0.4],
                  [0.9, 0.2, 0.6, 0.1],
                  [0.4, 0.4, 0.4, 0.4],
                  [0.0, 1.0, 0.0, 1.0]])
print(x.shape)        # 形状：5行4列
print(x.dtype)        # 数据类型
print(x.mean(dim=0))  # 按列求平均（dim=0 表示"沿着行的方向压缩"）
```
