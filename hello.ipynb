```python
dna = "ATGCGTACGTTAG"
complement = {"A": "T", "T": "A", "G": "C", "C": "G"}
result = "".join(complement[base] for base in dna)
print(result)   # 预期: TACGCATGCAATC
```

```python
gc = (dna.count("G") + dna.count("C")) / len(dna)
print(f"GC含量: {gc:.2%}")   # 预期: GC含量: 46.15%
```

```python
codons = [dna[i:i+3] for i in range(0, len(dna)-2, 3)]
print(codons)   # 预期: ['ATG', 'CGT', 'ACG', 'TTA']
```
